```python
import time
import numpy as np
import sys
```

```python
nums = [i for i in range(1, 11)]
large_num_array = nums * 1_000_000
```

```python
from copy import copy

new_nums = copy(large_num_array)
without_numpy_start = time.time()
for index in range(len(new_nums)):
    new_nums[index] = new_nums[index] * 2
without_numpy_end = time.time()

print(
    f"Time taken to perform the list multiplication without numpy: {without_numpy_end - without_numpy_start} and size of the array is {sys.getsizeof(nums)}"
)
```

```python
with_numpy_start = time.time()
numpy_nums = np.array(large_num_array)
new_numpy_nums = numpy_nums * 2  # Direct operation or vectorization
with_numpy_end = time.time()

print(
    f"Time taken to perform the list multiplication with numpy: {with_numpy_end - with_numpy_start} and size of the array is {np.array(nums).nbytes}",
)
```
